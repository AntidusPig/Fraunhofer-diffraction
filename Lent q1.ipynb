```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import linregress
```

```python
V1=[0.253,0.305,0.358,0.410]
d1=[28,14,7,3]
plt.scatter(d1,V1)
slope1,intercept1,_,_,_ = linregress(d1,V1)
x=np.array([0,30])
y=x*slope1+intercept1
plt.plot(x,y)
plt.scatter(0,intercept1)
plt.annotate("(0,$V_0$)",(0,intercept1))
plt.xlabel('$d$ ∝ number of electrons')
plt.ylabel('V')
print(intercept1)
```

```python
Vs = [
    [0.253,0.305,0.358,0.410],
    [0.829,0.881,0.934,0.986],
    [0.934,0.986,1.039,1.091,1.143],
    [1.353,1.405,1.458,1.510,1.562],
    [1.929,1.981,2.034,2.086,2.138]
]

ds = [
    [28,14,7,3],
    [44,20,10,4],
    [82,55,36,24,3],
    [67.5,36,19,11,4],
    [52,29,12,5,2.5]
]
```

```python
fig,axs = plt.subplots(3,2)
axs = axs.flatten()
fig.delaxes(axs[-1])
fig.tight_layout()
V0s = []
for V1,d1,ax in zip(Vs, ds, axs):
    ax.scatter(d1,V1)
    slope1,intercept1,_,_,_ = linregress(d1,V1)
    x=np.array([-1,d1[0]*1.1])
    y=x*slope1+intercept1
    ax.plot(x,y)
    ax.scatter(0,intercept1)
    V0s.append(intercept1)
    ax.annotate("(0,$V_0$)",(0,intercept1))
    ax.set_xlabel('d')
    ax.set_ylabel('V')
    print(intercept1)
```

```python
lamb = np.array([546,434,405,365,313])*1e-9
invlamb = 1/lamb
slope,intercept,_,_,_ = linregress(invlamb,V0s)
x=np.array([1.5e6,3.5e6])
y=x*slope+intercept
plt.scatter(invlamb, V0s)
plt.xlabel(r'$1/\lambda$')
plt.ylabel(r'$V_0$')
plt.plot(x,y)
print(f'slope={slope}')
print(f'intercept={intercept}')
```
